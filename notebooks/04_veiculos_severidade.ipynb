## Veículos envolvidos

```python
veic = df[VEICULOS].sum().sort_values(ascending=False)
ax = veic.plot(kind="barh", figsize=(8, 4), color="steelblue")
ax.invert_yaxis(); ax.set_title("Veículos envolvidos em acidentes (soma 2015-2024)")
plt.tight_layout(); plt.savefig(FIGURAS / "veiculos_envolvidos.png", dpi=150); plt.show()
veic
```

## Qual veículo está mais ligado a acidentes graves?

```python
res = {}
for v in VEICULOS:
    com = df[df[v] > 0]
    if len(com) > 0:
        res[v] = {"acidentes": len(com), "pct_com_morte": com["GRAVE"].mean() * 100}
pd.DataFrame(res).T.sort_values("pct_com_morte", ascending=False).round(2)
```
