```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))  # pra achar a pasta src

import pandas as pd
import geopandas as gpd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.config import *

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 40)
```

```python
df = gpd.read_parquet(ARQ_LIMPO)
df.shape
```

## Por ano

```python
por_ano = df.groupby("ANO").size()
ax = por_ano.plot(kind="bar", figsize=(9, 4), color="steelblue")
ax.set_title("Acidentes por ano"); ax.set_ylabel("quantidade")
plt.tight_layout(); plt.savefig(FIGURAS / "acidentes_por_ano.png", dpi=150); plt.show()
por_ano
```

> 2020 e 2021 costumam cair por causa da pandemia. Vale comentar isso no relatório.

```python
por_mes = df.groupby("MES").size()
por_mes.plot(kind="bar", figsize=(9, 4), color="steelblue", title="Acidentes por mês (todos os anos)")
plt.tight_layout(); plt.show()
```

## Por dia da semana

```python
por_dia = df["DIA_SEMANA"].value_counts().reindex(DIAS_SEMANA)
ax = por_dia.plot(kind="bar", figsize=(9, 4), color="steelblue")
ax.set_title("Acidentes por dia da semana"); ax.set_ylabel("quantidade")
plt.tight_layout(); plt.savefig(FIGURAS / "acidentes_por_dia_semana.png", dpi=150); plt.show()
por_dia
```

## Por hora

```python
por_hora = df.groupby("HORA_NUM").size()
ax = por_hora.plot(kind="line", marker="o", figsize=(9, 4))
ax.set_xticks(range(24)); ax.set_title("Acidentes por hora do dia"); ax.set_ylabel("quantidade")
plt.tight_layout(); plt.savefig(FIGURAS / "acidentes_por_hora.png", dpi=150); plt.show()
```

## Dia da semana x hora

```python
tab = df.pivot_table(index="DIA_SEMANA", columns="HORA_NUM", values="ANO", aggfunc="count").reindex(DIAS_SEMANA)
plt.figure(figsize=(12, 4))
sns.heatmap(tab, cmap="YlOrRd")
plt.title("Acidentes por dia da semana e hora"); plt.xlabel("hora"); plt.ylabel("")
plt.tight_layout(); plt.savefig(FIGURAS / "heatmap_dia_hora.png", dpi=150); plt.show()
```
